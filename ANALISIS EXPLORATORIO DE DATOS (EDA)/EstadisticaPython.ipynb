```python
from IPython.terminal.embed import line_magic
##  Ejemplo clase  Evolucion del total de ventas. usando grafico de linea
#   Id de Venta. Agrupar
#   Total. Variable


evolucion_ventas = df.groupby("ID_Venta")["Total_Venta"].sum()

evolucion_ventas.plot(
    kind="line",
    figsize=(10, 5),
    color="steelblue",
    marker="o",
    linestyle="--"
)

plt.title("Evolución del total de ventas")
plt.xlabel("ID de venta")
plt.ylabel("Total de ventas")
plt.show()
```

```python
# Crear total de cada venta
# Agrupar ventas por vendedor
# Crear gráfico

df["Total"] = df["Cantidad"] * df["Precio_Unitario"]

ventas_por_vendedor = df.groupby("Vendedor")["Total"].sum()

#Grafico en una variable
ax = ventas_por_vendedor.plot(
    kind="bar",
    figsize=(10, 5),
    color="steelblue"
)

#Mostrar total sobre la barra
plt.bar_label(ax.containers[0], labels=[f"₡{v:,.0f}" for v in ventas_por_vendedor])

#Mostrar total en el eje y
ax.yaxis.set_major_formatter('₡{x:,.0f}')

plt.title("Ventas totales por cada Vendedor")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
