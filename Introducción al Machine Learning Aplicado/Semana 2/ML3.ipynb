```python
import pandas as pd
import io
from google.colab import files
import numpy as np
import matplotlib.pyplot as ptl
from sklearn.linear_model import LinearRegression                               #Modelo de regresion lineal (ejecicios 1 y 2)
from sklearn.tree import DecisionTreeClassifier, plot_tree                      #Arbol de desicion

print("Entorno listo")

print("Selecciona el archivo")
subido = files.upload()
nombre_archivo = list(subido.keys())[0]
datos = pd.read_excel(io.BytesIO(subido[nombre_archivo]))
print(f"Archivo '{nombre_archivo}' cargado correctamente ")

 #datos = pd.DataFrame({"horas_estudio":[1,3,5,7,9],                            #columna con las horas que estudio[cite: 1]
                       #"nota":[46,58,70,82,94],                                #columa con la nota que obtuvo cada estudiante (0-100)[cite: 1]
                       #})

print(f"Archivo '{nombre_archivo}' cargado correctamente ")
print(datos.head())

X=datos[["horas_estudio"]]                                                      #x va con doble corchete porque es una tabla
y=datos["nota"]                                                                 #y es en solo corchete porque es una columna


#Creamos el modelo de regresion lineal
modelo=LinearRegression()

#Entrenamos el modelo
modelo.fit(X,y)
print("Modelo entrenado y listo...EXACTO GINEEEETH")                            #Confirmamos que el entrenamiento termino sin errores

#Revision de que aprendio el modelo
pendiente=modelo.coef_[0]                                                       #coef_ guarda cuanto cambia "Nota" por cada hora de estudio
base=modelo.intercept_                                                          #intercept_ es la nota que el modelo espera con 0 horas de estudio

print(f"Por cada hora de estudio, la nota sube unos {pendiente:.1f} puntos")    # interpretamos la pendiente
print(f"Con 0 horas de estudio, el modelo espera una nota base de {base:.1f}")  # interpretamos la base

#Probamos el modelo con un valor nuevo
horas_nuevas=pd.DataFrame({"horas_estudio":[8]})                                #usamos un DF con el mismo nombre de la columna que x
prediccion=modelo.predict(horas_nuevas)                                         #le pedimoos al modelo que prediga la nota para ese valor

print(f"Prediccion para 6 horas de estudio: {prediccion[0]:.1f} puntos")        #Mostrar el resutado con un decimal

```

```python
#Creamos la rabla con 4 telefonos ya vendidos
telefonos=pd.DataFrame({
    "meses_uso":[3,18,30,8],              #Antiguedad del telefono
    "bateria_pct":[95,82,70,90],          #Porcentaje de bateria
    "precio":[420,260,150,380]            #Precio en dolares
})

print(telefonos)                          #Mostrar la tabla para revisar los datos

# X son las pistas: las columnas que el modelo usa para predecir
X=telefonos[["meses_uso","bateria_pct"]]  #dos columnas ---> va en doble corchete

# y es lal rrspuesta: la columna que queremos calcular
y=telefonos["precio"]

#Creamos y entrenamos el modelo con las pista a la vez
modelo_telefonos=LinearRegression()
modelo_telefonos.fit(X,y)
print("Modelo de telefonos listo y entrenado")

#Probamos el modelo con un telefono nuevo 12 meses y un 85% de bateria
telefono_nuevo = pd.DataFrame({"meses_uso": [12], "bateria_pct": [85]})

prediccion=modelo_telefonos.predict(telefono_nuevo)

print(f"Precio estimado para el teléfono nuevo: ${prediccion[0]:.2f}")
```
